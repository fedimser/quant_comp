## IQM demo

In this notebook I run several very simple quantum circuits on IQM devices to get a feel of running algorithms on real quantum hardware and also understand how large are the errors.

For each circuit, I first simulate it on simulator, then submit to all 3 available machines ("garnet" was unavailable for some runs).

This is based on IQM's [tutorial](https://github.com/d-bulmash/ieee-qec-tutorial-2026).

## Exercise 1: Applying X gate on each qubit and measuring (with 10 qubits).

```python
reg1 = qrisp.QuantumVariable(10)
qrisp.x(reg1)
qrisp.measure(reg1)
reg1.get_measurement(shots=100)
```

```python
reg1.get_measurement(backend=devices["emerald"],shots=1000)
```

```python
reg1.get_measurement(backend=devices["sirius"],shots=1000)
```

## Exercise 2: GHZ state preparation on 10 qubits

```python
n = 10
reg2 = qrisp.QuantumVariable(n)
qrisp.h(reg2[0])
for i in range(1, n):
    qrisp.cx(reg2[0], reg2[i])
qrisp.measure(reg2)
reg2.get_measurement(shots=100)
```

```python
reg2.get_measurement(backend=devices["emerald"],shots=1000)
```

```python
reg2.get_measurement(backend=devices["sirius"],shots=1000)
```

## Exercise 3: Bernstein-Vazirani algorithm on 10 qubits

See https://github.com/microsoft/qdk/blob/main/samples/algorithms/BernsteinVazirani.qs

```python
def apply_parity_operation(s, x_register, y_qubit):
    n = s.bit_length()
    for i in range(n):
        if (s&(1<<i)):
            qrisp.cx(x_register[i], y_qubit)
    
def bernstein_vazirani(s):
    n = s.bit_length()
    query_register = qrisp.QuantumVariable(n)
    target = qrisp.QuantumVariable(1)
    qrisp.x(target)
    qrisp.h(target)
    qrisp.h(query_register)
    apply_parity_operation(s, query_register, target)
    qrisp.h(query_register)
    measure(query_register)
    return query_register
    
s = 997        
reg3 = bernstein_vazirani(s)
reg3.get_measurement(shots=1000)
```

```python
reg3.get_measurement(backend=devices["garnet"],shots=1000)
```

```python
reg3.get_measurement(backend=devices["emerald"],shots=1000)
```

```python
reg3.get_measurement(backend=devices["sirius"],shots=1000)
```

We recover the correct hidden string with probability 50-60%.
